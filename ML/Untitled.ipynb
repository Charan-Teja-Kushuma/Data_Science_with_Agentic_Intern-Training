```python
#Creating student dataset
import pandas as pd
from sklearn.model_selection import train_test_split

data = {
    "Hours":[2,3,4,5,6,7,8,9,4,6],
    "Attendance":[55,60,65,70,75,80,85,90,68,78],
    "Pass":[0,0,0,1,1,1,1,1,0,1]
}

df=pd.DataFrame(data)
```

```python
print(df)
```

```python
#wseparatee can separate Feature and Label
x = df[["Hours", "Attendance"]]
y = df[["Pass"]]
```

```python
#We can implement the train and test split of the data
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
print("Training Features")
print(x_train)
```

```python
print("\nTesting Feature")
print(x_test)
```

```python
print("\nTraining labels")
print(y_train)
```

```python
print("\nTesting labels")
print(y_test)
```
