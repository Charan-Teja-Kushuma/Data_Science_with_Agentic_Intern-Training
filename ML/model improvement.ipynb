```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


#create dataset
data = {
    "Age": [22, 25, 30, 35, 40, 28, 32, 45, 26, 38],
    "Salary": [
        25000, 35000, 50000, 75000, 100000,
        45000, 60000, 120000, 32000, 85000
    ],
    "Experience": [
        1, 3, 5, 8, 12,
        4, 6, 15, 2, 10
    ]
}

df = pd.DataFrame(data)

print ("dataset:", df)
```

```python
# select of the Feature of the dataset

x = df[["Age", "Salary", "Experience"]]


        
```

```python
#Train and Test split
X_train, X_test = train_test_split(
    x,
    test_size=0.2,
    random_state=42
)
```

```python
#creatng scaler
scaler = StandardScaler()

#fit and transform training

X_train_scaler = scaler.fit_transform(
    X_train
)
```

```python
#test the transform data
X_test_scaler = scaler.transform(
    X_test
)
```

```python
#display the result
print("\n Original Training Data:",X_train)

print("\n Scaler Training Data:",X_train_scaler)

print("\n Original Testing Data:",X_test)

print("\n Original Testing Data:",X_test_scaler)
```
